### Séries et DataFrames

On importe généralement le module Pandas de la manière suivante:
```python
import pandas as pd
```

Pandas fournit deux structures de données :
- `pd.DataFrame` représente une table de données, dans laquelle chaque colonne a un nom et chaque ligne est repérée par un index ;
- `pd.Series` représente une colonne de cette table.

__DataFrame__


Un objet de type `pd.DataFrame` représente une table dans laquelle chaque colonne a un nom et contient des objets d'un type donné (entier, réel, chaîne de caractères, etc.).

![dataframe](../images/dataframe.png)


On peut créer un DataFrame 
- soit à partir de listes de données :

````python
import pandas as pd
pd.DataFrame([[4, 7, 10],[5, 8, 11],[6, 9, 12]],index=[1, 2, 3],columns=['a', 'b', 'c'])
````

- soit à partir d'un tableau multidimensionnel (de type `ndarray`) :

````python
import numpy as np
data = np.array([[4, 7, 10],[5, 8, 11],[6, 9, 12]])
pd.DataFrame(data,index=[1, 2, 3],columns=['a', 'b', 'c'])
````

- soit à partir d'une dictionnaire :

````python
pd.DataFrame({'a':[4, 7, 10], 'b':[5, 8, 11], 'c':[6, 9, 12]},index=[1, 2, 3])
````

- soit à partir d'un fichier csv

````python
col_names = ['id',"hair", "feathers", 'eggs', 'milk', 'airborne', 'aquatic', 'predator','toothed', 'backbone', 'breathes', 'venomous', 'fins', 'legs', 'tails', 'domestic', 'catsize','type']
zoo = pd.read_csv('../data/zoo.data',index_col=0,names=col_names)
zoo
````

Les données d'un DataFrame sont stockées dans un `ndarray`:

````python
zoo.columns, zoo.index, zoo.values, type(zoo.values)
````

__Series__

Chaque colonne d'un DataFrame est un objet de type `pd.Series`, qui représente une série de données avec son nom de série et un index pour ses valeurs.

![series](../images/series.png)

Par exemple, la colonne "eggs" de la table `zoo` est une série de valeurs entières indexées par les noms d'animaux (qui est l'index de la table elle-même).

````python
s = zoo['eggs']
s, type(s), s.index, s.values
````

Pour créer une série, on crée un objet de type `pd.Series` en lui passant une séquence de valeurs et un index pour ces valeurs :

````python
pd.Series(np.random.randn(5), index=["a", "b", "c", "d", "e"])
````

<a id="index"></a>
### Index

Les lignes d'une table sont identifiées de manière unique à l'aide d'un index (qui joue un peu le rôle d'une clé primaire, comme dans le modèle relationel).

````python
df = pd.DataFrame({'month': [1, 4, 7, 10],'year': [2012, 2014, 2013, 2014],'sale': [55, 40, 84, 31]})
df, df.index
````

__Définir / modifier l'index__

On peut modifier un index à l'aide des fonctions 
[set_index](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.set_index.html#pandas.DataFrame.set_index)
et [reset_index](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.reset_index.html#pandas.DataFrame.reset_index)

La fonction `set_index` permet de choisir la (ou les) colonne(s) qui serviront d'index :

````python
df.set_index('month')
````

````python
df.set_index(['year','month'])
````

La fonction `reset_index` permet de réinitialiser l'index à l'index par défaut (numérotation 1, 2, 3...).

Par défaut, Pandas ajoute une colonne index :

````python
df.reset_index()
````

L'option `drop=True` permet de ne pas ajouter cette colonne :

````python
df.reset_index(drop=True)
````

La fonction renvoie la nouvelle DataFrame, sauf si l'option `inplace=True` est ajoutée. Dans ce cas, la modification de l'index se fait sur place, et la fonction renvoie None :

````python
df.reset_index(drop=True, inplace=True)
````

<a id="selection"></a>
### Sélection / projection

__Sélection par index__

Plusieurs méthodes peuvent être appelées sur un objet de type `pd.DataFrame` pour sélectionner certaines lignes, certaines colonnes, ou accéder à des valeurs par leur index :
- `df.iloc[]` permet de faire de la __sélection par numéro__ de ligne ou colonne :

````python
# la 4e ligne seulement
zoo.iloc[3]
````

````python
# les 5 premières lignes, pour les colonnes de eggs à aquatic
zoo.iloc[:5,2:6]
````

- `df.loc[]` permet de faire de la __sélection par label__ de ligne ou colonne

````python
# la même chose, mais sélection par label
zoo.loc[:'boar','eggs':'aquatic']
````

- `df.at` et `df.iat` permet d'accéder à une seule valeur étant donné son index

````python
# la valeur à la ligne 1 colonne 4
zoo.iat[1,4]
````

````python
# la même chose avec des labels
zoo.at['antelope','airborne']
````

__Requêtes__

Les méthodes `df.loc[]` et `df.query()` peuvent être utilisées pour sélectionner les lignes qui satisfont plusieurs conditions, à la manière d'une clause WHERE :

````python
zoo.loc[(zoo['legs'] == 4) & (zoo['aquatic'] == 1)]
````

````python
zoo.query('legs == 4 and aquatic == 1')
````

La méthode `df.loc[]` est généralement plus rapide, mais quelque fois moins lisible...

__Filtrer l'index (ou les noms de colonnes)__

La méthode `df.filter()` permet de filtrer l'index ou les noms de colonnes, par exemple pour 
- ne garder que les animaux qui commencent par `m` :

````python
zoo.filter(regex='^m', axis=0)
````

- ne garder que les attributs qui contiennent `size`:

````python
zoo.filter(regex='.*size',axis=1)
````

`df.head()` affiche les premières lignes du jeu de données :

Pour chaque attribut, on peut lister quelques statistiques (nombre de valeurs, moyenne, médiane, Q1, Q3, écart type, max) :

__Valeurs manquantes__

Les valeurs manquantes (notées `NaN` dans un tableau numérique) peuvent être détectées à l'aide de la fonction `pd.isna` :

````python
df.loc[1,'sale']=np.nan
df, pd.isna(df)
````

Il y a plusieurs moyens de traiter les valeurs manquantes :
- soit on enlève l'instance toute entière 

````python
df.dropna()
````

- soit on attribue la valeur moyenne sur le jeu de données

````python
df.fillna(df.mean(0)['sale'])
````

<a id="reorganiser"></a>
### Réorganiser les données

Pandas fournit de nombreuses fonctions pour [consolider, fusionner](https://pandas.pydata.org/pandas-docs/stable/user_guide/merging.html) les données, comme par exemple :
- `concat()` pour mettre bout à bout deux tables

````python
df1 = pd.DataFrame([[4, 7, 10],[5, 8, 11],[6, 9, 12]],index=[1, 2, 3],columns=['a', 'b', 'c'])
df1
````

````python
df2 = pd.DataFrame([[21,23,25]],index=[4],columns=['a', 'b', 'c'])
df2
````

````python
pd.concat([df1,df2])
````

- `join` pour faire un jointure (comme en SQL), mais sur les index seulement :

````python
d_df = pd.DataFrame([[21],[8]],index=[1,2],columns=['d'])
d_df
````

````python
df1.join(d_df)
````

- `merge` pour faire aussi des jointures sur les colonnes (comme ici sur la colonne `b`)

````python
d_df = pd.DataFrame([['ours'],['caribou']],index=[7,9],columns=['label for b'])
d_df
````

````python
pd.merge(df1,d_df,left_on='b',right_index=True)
````

- `assign()` pour créer une nouvelle colonne et la remplir

````python
df1.assign(d=lambda some_df: some_df.a + 4) # ou df.assign(**{'d':(lambda some_df: some_df.a + 4)})
````

Pandas fournit aussi des fonctions pour [résumer ou réorganiser les données](https://pandas.pydata.org/pandas-docs/stable/user_guide/reshaping.html), comme par exemple :
- `pivot_table` qui produit un tableau croisé

````python
df = pd.DataFrame({'id':['E'+str(i) for i in range(1,7)],
                   'groupe':['A','A','B','A','B','A'],
                   'note':[3,18,9,1,19,10]}).set_index('id')
df
````

````python
pd.pivot_table(df,index='groupe',values='note',aggfunc='mean')
````

- ou `groupby` qui permet de regrouper les lignes selon les modalités d'une variable avant d'appeler une fonction d'aggrégation, comme en SQL :

````python
df.groupby('groupe')['note'].max()
````
