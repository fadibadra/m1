<a id="vues"></a>
##### Vues

Il y a plusieurs moyens d'[accéder aux éléments](https://numpy.org/doc/stable/user/basics.indexing.html) d'un tableau.

```python
import numpy as np
a = np.array([[1, 2], [3, 4], [5, 6]])
a
```

__Accéder à un élément__

Pour accéder à la valeur à la $3^e$ ligne, $1^{ère}$ colonne : 

```python
a[2,0] # ou a[2][0] 
```

<div><img src="../images/indexation_simple.png" width="600" /></div>


De la même façon :

```python
a[1]
```

Il faut noter que cette syntaxe permet d'accéder à une __vue__ sur le tableau, pas à une copie du tableau (si on modifie la vue, le tableau original est lui aussi modifié).

On peut aussi accéder aux éléments passant des listes d'indices pour chaque dimension :

```python
a[[0,2],[1]]
```

<a id="slicing"></a>
##### Slicing

On utilise la même syntaxe que lorsqu'on manipule des séquences en Python pour accéder à des sous-séquences.

Par exemple, si on veut accéder aux deux premières lignes et à la première colonne :

```python
a[:2,0]
```

<a id="copies"></a>
##### Copies

On peut demander à créer une copie du tableau avec la méthode `copy()`.

```python
b = a.copy()
a[2]= np.array([[0,9]])
a,b
```

<a id="tris"></a>
##### Tris et filtres

NumPy fournit plusieurs fonctions pour faire des [tris et recherches](https://numpy.org/doc/stable/reference/routines.sort.html#sorting) dans un tableau.

Voici quelques exemples.

__Recherche de max ou min__

La fonction `np.max` renvoie le plus grand élément du tableau qu'on lui passe comme argument
et `np.argmax` l'indice de cet élément.

```python
a = np.array([[1, 2], [3, 4], [5, 0]])
a
```

```python
np.max(a, axis=1)
```

```python
np.argmax(a, axis=1)
```

__Tris__

La fonction `np.sort()` implémente des algorithmes de tri. 

```python
from cbp.similarity.measures import euclidean_sim

similarities = [euclidean_sim((0,3),xi) for xi in [(2,-5),(8,-4),(-17,-23),(28,0),(-8,-7),(3,2.9),(23,14),(7,14),(-2,28)]]
np.sort(similarities)
```

On peut récupérer les indices des éléments triés :

```python
np.argsort(similarities)
```

__Valeurs uniques__

La fonction `np.unique()` permet d'enlever les doublons dans un tableau.

```python
y = np.array([1,0,1,0,1,1,0,0,1])
np.unique(y)
```

On peut récupérer les indices de chaque valeur du tableau initial dans le tableau de valeurs uniques :

```python
b = np.array([4.,8.,1.,4.,1.])
u, indices = np.unique(b,return_inverse=True)
u, indices, u[indices]
```

On peut aussi récupérer le nombre de fois où chaque valeur apparaît :

```python
b = np.array([4.,8.,1.,4.,1.])
u, counts = np.unique(b,return_counts=True)
u, counts
```

__Filtres__

La fonction `np.where()` permet de sélectionner les lignes d'un tableau qui satisfont certaines conditions.

```python
np.where(b<5)
```
