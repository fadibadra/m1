<a id="noms"></a>
### Les espaces de noms

Un __espace de noms__ est un ensemble de correspondances faites
entre des noms et des objets.
Python crée un nouvel espace de noms pour chaque fonction, module,
paquetage, classe.

Techniquement, un espace de noms est représenté par Python par
un dictionnaire qui associe des objets à des noms.
La fonction `vars()` renvoie ce dictionnaire.

Voici un programme Python :

```python
a = 3
def f():
    b = 2
    import re
    a = 7 
    def g(x):
        return x**2 + b
    a = g(a) 

f()
print(a)
```

L'instruction suivante affiche (joliment) l'espace de noms du module `__main__` dans lequel on se trouve par défaut :

```python
import __main__
from pprint import pprint as pretty_print
pretty_print(vars(__main__))
```

Remarque : on peut aussi utiliser la fonction `globals()` pour accéder à l’espace de noms du module `__main__`.

```python
globals()
```

Afficher de même les espace de noms des modules suivants :
- le module `__builtins__`
- le module `re`

Pour afficher l'espace de noms des fonctions `f()` et `g()`, on peut appeler la fonction `vars()` dans le corps de la fonction :

```python
a = 3
def f():
    b = 2
    import re
    a = 7 
    print('portee locale de f', vars())
    def g(x):
        print('portee locale de g', vars())
        return x**2 + b
    a = g(a) 

f()
```

Remarque : on peut aussi utiliser la fonction `locals()` pour accéder à l’espace de noms d’une fonction.

### La visibilite des espaces de noms

Tous les espaces de noms ne sont pas forcément visibles au cours de l'exécution d'un programme.
A un instant $t$ à l'exécution, un espace de noms est soit :
- __directement accessible__ : les noms de cet espace peuvent être utilisés directement
- __indirectement accessible__ : les noms peuvent être utilisés, mais en spécifiant d'où ils viennent, avec la notation pointée
- __invisible__




On appelle __portée__ un espace de noms directement accessible
à un instant donné de l'exécution.

A l'exécution, Python gère plusieurs portées :
- la portée __locale__ = les noms d\'efinis dans le
    *code block* courant ;
- les port\'ees des éventuelles fonctions appelantes ;
- la portée __globale__ = du module dans lequel on
    se trouve ;
- la portée du module `builtins`, toujours directement accessible.

Ces portées sont imbriquées :

__locale__ $\subseteq$ fonctions appelantes $\subseteq$  __globale__ $\subseteq$ __builtins__

Lorsque Python fait la résolution d'un nom,
il cherche d'abord dans la portée locale,
puis remonte les portées imbriquées jusqu'à ce qu'il trouve le nom.
Si le nom n'est pas trouvé, une erreur de type `NameError` est levée.

Par exemple, dans le programme suivant :
- la fonction `print()` définie dans le module `__builtins__` est directement accessible depuis le module `__main__` où l'on se trouve
- la fonction `g()` est directement accessible depuis le corps de la fonction `f()`, mais invisible en dehors
- la fonction `search()` du module `re` est indirectement accessible depuis le corps de la fonction `f()`, mais invisible en dehors.

```python
a = 3
def f():
    b = 2
    import re
    re.search(r'C+','GAAACCCG')
    a = 7 
    def g(x):
        return x**2 + b
    a = g(a) 

f()
print(a)
g()
re.search(r'A+','GAAAC')
```

Supposons maintenant qu'à la ligne 3 de notre programme, 
on rajoute l'instruction `global a`
qui spécifie que le symbole `a` utilisé dans la fonction `f()` n'est pas une variable à créer dans sa portée locale, mais correspond au nom `a` précédemment défini dans la portée globale :

```python
a = 3
def f():
    global a
    b = 2
    import re
    a = 7 
    def g(x):
        return x**2 + b
    a = g(a) 

f()
print(a)
```
