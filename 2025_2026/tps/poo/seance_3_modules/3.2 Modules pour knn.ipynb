### Le module knn

Dans le paquetage `cbp.algo`, ajouter un module `knn` dans lequel on placera les fonctions `similarites`, `tri_bulle`,`vote_majoritaire` et `predict` écrites au 2.3.

Maintenant, on doit pouvoir utiliser la fonction `predict` pour prédire le label associé à une nouvelle entrée, comme ici :

```python
from cbp.similarity import euclidean_sim
from cbp.algo import knn

# la base de cas
X = [(2,-5),(8,-4),(-17,-23),(28,0),(-8,-7),(3,2.9),(23,14),(7,14),(-2,28)]
y = [1,0,1,0,1,1,0,0,1]

# la nouvelle entrée
x = (0,3)

knn.predict(x, X, y, euclidean_sim, n_neighbors=5)
```

Implémenter cette fonction pour qu'elle renvoie la liste des indices des cas de `X_cb` les plus similaires au nouveau cas `x` :

```python
knn.kneighbors(x, X, euclidean_sim, n_neighbors=3)
```
