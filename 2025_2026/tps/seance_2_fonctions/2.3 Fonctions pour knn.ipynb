- divisant le calcul compliqué en plusieurs calculs plus simples ;
- donnant un nom à chaque sous-calcul et on le met dans une fonction ;
- combinant ces fonctions pour réaliser le calcul compliqué.

Ici, on peut décomposer l'algorithme $k$-NN plusieurs sous-calculs :
- initialisation

    - la définition de la base de cas
- étape 1 : calcul des similarités
- étape 2 : tri des similarités et sélection des k plus similaires
- étape 3 : vote majoritaire
